# Problem 153. Create, Access & Update a Dictionary
Create a dictionary describing a person, read one value, change an
existing value, and add a new key.

```python
person = {"name": "yogesh", "age":25}
print(person["name"])          #-----Access by key
person["age"]= 26        #-----Update existing key
person["city"]= "pune"      #-----add new key
print(person)
```

# Problem 154. Iterate Keys, Values & Items
Loop over a dictionary three ways: keys only, values only, and key
value pairs together.

```python
marks= {"math": 75, "science": 81}
for subject in marks.keys():       #------Loops over keys by default 
    print(subject)

for scores in marks.values():        #----Loops over values by default
    print(scores)

for subject,score in marks.items():
    print(f"{subject} = {score}")
```

# Problem 155. Safe Key Lookup with .get()
Look up keys that may not exist without crashing, and provide a
default value for missing keys.

```python
person = {"name": "Ravi", "age": 30}

print(person.get("name"))         #----Key exisit      
print(person.get("city"))       #---- missing None , no crash

print(person.get("city", "Not found"))  # missing -> custom default

# person["city"] would raise KeyError instead
print("city" in person)       #---- membership test

```

# Problem 156. Sum of Dictionary Values
A shopping cart maps item names to prices. Compute the total bill
and the number of items.

```python
cart = {"apple": 30, "milk": 55, "bread": 40}

print("items:", len(cart))

print("total:", sum(cart.values()))
```

# Problem 157. Key with the Maximum Value
Given student scores in a dictionary, find which student scored the
highest.

```python
scores = {"Amit": 82, "Bela": 91, "Chad": 77}

best= max(scores, key=scores.get)      #---- compare keys by their values, ---Ye line dictionary ki keys ko compare karegi, unki values ke basis par

print(f"{best} scored the highest: {scores[best]}")
```

# Problem 158. Dict from Two Lists via zip()
Build a dictionary from a list of names and a matching list of ages.

```python
names = ["Asha", "Ravi", "Meena"]
ages = [25, 30, 28]

people= dict(zip(names,ages))       #----# zip pairs items position by position
print(people)
```

# Problem 159. Dictionary Comprehension
Build a dictionary mapping each number from 1 to 5 to its square
using a dict comprehension.

```python
squares = {n: n * n for n in range(1, 6)}
print(squares)

# with a condition: only even keys
even_squares = {n: n * n for n in range(1, 11) if n % 2 ==0}
print(even_squares)
```

# Problem 160. Character Count with a Dict
Count how many times each character appears in a word, using a
plain dictionary.

```python
word = "banana"
counts = {}
for ch in word:
    counts[ch] = counts.get(ch, 0) + 1   # 0 if seen first time

print(counts)
```

# Problem 161. Word Frequency Counter
Count how many times each word appears in a sentence.

```python
text = "to be or not to be"
freq = {}

for word in text.split():
    freq[word] = freq.get(word, 0) + 1
print(freq)
```

# Problem 162. Invert a Dictionary
Swap the keys and values of a dictionary (assume all values are
unique).

```python
capitals = {"India": "Delhi", "Japan": "Tokyo"}

inverted = {city: country for country, city in capitals.items()}
print(inverted)
```
