# Problem 116. Maximum Without max()
Find the largest number in a list without using the built-in 
function.

```python
numbers = [12, 45, 2, 67, 33]
largest = numbers[0]

for n in numbers:
    if n > largest:
        largest= n
print(largest)
```

# Problem 119. Linear Search
Search for a target value in a list and print the index of its first
occurrence, or -1 if it is not present.

```python
numbers = [4, 7, 1, 9, 3]
target = 9
index = -1
for i in range(len(numbers)):
    if numbers[i] == target:
        index = i
print("Index of", target, ":", index)
```

# Problem 120. Count Occurrences of a Value
Count how many times a given value appears in a list, without
using 
list.count().

```python
numbers = [1, 2, 3, 2, 4, 2]
target = 2
count = 0
for n in numbers:
    if n == target:
        count += 1
print(target, "appears", count, "times")
```

# Problem 122. Reverse a List In Place
Reverse a list in place by swapping elements, without using
reverse() or slicing.

```python
numbers = [1, 2, 3, 4, 5]
left = 0
right = len(numbers) - 1
while left < right:
    numbers[left], numbers[right] = numbers[right], numbers[left]
    left += 1
    right -= 1
print("Reversed:", numbers)
```

# Problem 123. Check If a List Is Sorted
Check whether a list is sorted in non-decreasing order.

```python
def is_sorted(nums):
    for i in range(len(nums) - 1):
        if nums[i]>nums[i+1]:
            return False
    return True

print(is_sorted([1,5,6,7]))
is_sorted([1,7,6,8,7])
 
```

# Problem 124. Second Largest Element
Find the second largest distinct value in a list without sorting it.

```python
nums= [10,50,6,0,20,30]
largest= None
second= None
for n in nums:
    if largest is None or n > largest:
        second= largest
        largest= n 
    elif n != largest and (second is None or n> second):
        second= n
print("second largest number is:", second)
```

# Problem 125. Remove Duplicates (Keep Order)
Remove duplicate values from a list while keeping the original order
of first appearances.

```python
nums= [1,5,7,8,5,3,54,1]
unique=[]
for i in nums:
    if i not in unique:
        unique.append(i)
print(unique)
```
