Problem 48. Simple Calculator Using match-case
Read two numbers and an operator (+, -, *, /) and print the result using Python's match-case
statement. Guard against division by zero

```python
a= int(input("enter operator"))
b= int(input("enter operator"))
op = input("enter operator")

match op:
    case "+":
        print("result: ", a+b)
    case "-":
            print("result: ", a-b)
    case "*":
            print("result: ", a*b)
    case "/":
        if b==0:
            print("can not divide by zero")
        else:
            print("result: ", a/b)
    case "_":
          input("invalid operator")
```

Problem 50. Income Tax Calculator with Slabs
Compute income tax: no tax up to Rs. 2,50,000; 5% on the next Rs.
2,50,000; 20% on the next Rs. 5,00,000; and 30% on income
above Rs. 10,00,000. Print the total tax

```python
income = float(input("Enter annual income: "))
tax = 0.0
if income > 1000000:
    tax += (income - 1000000) * 0.30   
    income = 1000000                   

if income > 500000:
    tax += (income - 500000) * 0.20    
    income = 500000
    
if income > 250000:
    tax += (income - 250000) * 0.05    

print(f"Total tax = Rs. {tax:.2f}")
```
