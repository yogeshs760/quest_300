# Problem 23. Sum of Digits

```python
num = int(input("Enter a number: "))

total= 0
n = num
while n > 0:
    total += n % 10   # n % 10 gives the last digit
    n //= 10          

# n //= 10 removes the last digit
print("Sum of digits of", num, "is", total)
```

# Problem 26. Power Without Using **
Read a base and an exponent, and compute base raised to the
exponent using a loop instead of the ** operator.

```python
base = int(input("enter base: "))
exp = int(input("enter power: "))

result = 1 
for _ in range(exp):
    result *= base
print(f"power {exp} of base {base} is: {result}")
```

# Problem 27. Palindrome Number
Read a number and check whether it reads the same backwards
(e.g., 121 is a palindrome, 123 is not).

```python
num = int(input("enter value:"))
rev = 0
n = num
while n > 0:
    rev = (rev * 10) + (n % 10)
    n = n//10

if rev == num:
    print(num," number is palindrome")
else:
    print(num, "is not palindrome ")
```

# Problem 30. Prime Number Check
Read a number and check whether it is prime. A prime number has
exactly two factors: 1 and itself.

```python
num = int(input("Enter a number: "))
is_prime = num > 1

for i in range(2, int(num**0.5) + 1):
    if num % i == 0:
        is_prime = False
        break  

if is_prime:
    print(num, "is a prime number")
else:
    print(num, "is not a prime number")
```

# Problem 31. Armstrong Number


```python
num = int(input("Enter a number: "))
digits = len(str(num))   # how many digits the number has
total = 0
n = num
while n > 0:
    total += (n % 10) ** digits
    n //= 10

if total == num:
    print(num, "is an Armstrong number")
else:
    print(num, "is not an Armstrong number")
```

# Problem 32. Perfect Number Check


```python
num = int(input("Enter a number: "))
total= 0
n= num
for i in range(1 , num):
    if num % i ==0:
        total += i
if total == num:
    print(num,"is a perfect number")
else:
    print(num, "is not a perfect number")
```

# Problem 33. GCD of Two Numbers
Read two numbers and find their greatest common divisor (GCD)
using the Euclidean algorithm.

```python
num = int(input("Enter a number: "))
num2 = int(input("Enter a number: "))

x,y = num, num2
while y !=0:
    x,y=y, x % y

print(f"{num} and {num2} GDC is {x}")
```

# Problem 34. LCM of Two Numbers
Read two numbers and find their least common multiple (LCM)
using the relation LCM(a, b) = a × b / GCD(a, b).

```python
num = int(input("Enter a number: "))
num2 = int(input("Enter a number: "))

x,y = num, num2
while y !=0:
    x,y=y, x % y

lcm = num * num2 // x
print(f" LCM of {num} and {num2} is {lcm}")
```

# Problem 35. Roots of a Quadratic Equation
Read coefficients a, b, c of ax² + bx + c = 0 and print its roots using the quadratic formula.
Handle real-and-distinct, real-and-equal, and complex cases

```python
import math
a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))
d = b ** 2 - 4 * a * c   # discriminant decides the type of roots

if d > 0:
    root1 = (-b + math.sqrt(d)) / (2 * a)
    root2 = (-b - math.sqrt(d)) / (2 * a)
    print(f"Two real roots: {root1} and {root2}")
elif d == 0:
    root = -b / (2 * a)
    print(f"One repeated root: {root}")
else:
    real = -b / (2 * a)
    imag = math.sqrt(-d) / (2 * a)
    print(f"Complex roots: {real} + {imag}i and {real} - {imag}i")
```
